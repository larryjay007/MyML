```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)
visible_pages = (df["impressions_90d"] > 0).sum()
declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = declining_pages / total_pages

print(f"Total pages in starter dataset: {total_pages}")
print(f"Pages with measurable impressions (impressions_90d > 0): {visible_pages}")
print(f"Pages currently flagged as declining (trend_direction == 'down'): {declining_pages}")
print(f"Share of all pages that are declining: {declining_pct:.1%}")
```

Of the 30,000 pages in the starter dataset, 100% have measurable impressions in the last 90
days — this dataset is pre-filtered to visible pages only, so there's no "zero-traffic" noise
to filter out. More importantly: 16,262 pages (54.2%) are currently flagged as declining.
That's over half the inventory — far more than any small team could manually review one by one.
This is exactly the capacity problem Lane 2 addresses: not "is there a problem," but "which of
these 16,000+ pages should a reviewer look at first."

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What this work can say: this is an observed pattern in a 30,000-page anonymized sample — over
half the pages show a current-window decline flag. Any ranking I build will be directional and

What this work can never say: it cannot claim to have proven why any page is declining, cannot
